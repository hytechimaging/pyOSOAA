### <span style="color:#D2383F">class SEA</span>

Définition des interfaces au fond de l'océan et avec l'air.


* <span style="color:#2882E2">surfalb</span> : Réflexion lambertienne de l'écume à la longueur d'onde de calcul. *Valeur par défaut : <span style="color:#009600">0.0</span>*. *Résolution : 0.001*. *Plage de valeurs : [0.0,1.0]*.

* <span style="color:#2882E2">bottype</span> : Type de fond pour la définition de l'albédo. (1) valeur lambertienne donnée par l'utilisateur dans <span style="color:#2882E2">SEA.botalb</span>, (2) sable clair, (3) algues vertes, (4) algues brunes, (5) algues rouges.  *Valeur par défaut : <span style="color:#009600">1</span>*

* <span style="color:#2882E2">botalb</span> : Albédo du fond de l'océan pour la longueur d'onde de calcul. *Valeur par défaut : <span style="color:#009600">0.30</span>*. *Résolution : 0.001*. *Plage de valeurs : [0.0,1.0]*.

* <span style="color:#2882E2">wind</span> : Vitesse du vent (m/s). *Valeur par défaut : <span style="color:#009600">7</span>*. *Résolution : 0.001 m/s*. *Valeur min : 0 m/s*.

* <span style="color:#2882E2">ind</span> : Indice de réfraction surface/atmosphère (air = 1). *Valeur par défaut : <span style="color:#009600">1.34</span>*. *Résolution : 0.001*. *Valeur min : 0*.

* <span style="color:#2882E2">depth</span> : Profondeur de l'océan (m). Si aucune valeur n'est renseignée, la profondeur euphotique d'après les données de concentration en chlorophylle de Morel est utilisée. *Valeur par défaut : <span style="color:#009600">15.0</span>*. *Résolution : 0.001 m*. *Valeur min : 0 m*.

```python
atlantic_ocean = pyOSOAA.osoaa.SEA()
atlantic_ocean.surfalb = 0.56
atlantic_ocean.bottype = 3
atlantic_ocean.botalb = 0.002
atlantic_ocean.wind = 12
atlantic_ocean.ind = 1.35
atlantic_ocean.depth = 50
```

```python
print("surfalb = ", atlantic_ocean.surfalb)
print("bottype = ", atlantic_ocean.bottype)
print("botalb = ", atlantic_ocean.botalb)
print("wind = ", atlantic_ocean.wind)
print("ind = ", atlantic_ocean.ind)
print("sdepth = ", atlantic_ocean.depth)
```

### <span style="color:#D2383F">class LOG</span>

Définition des noms de fichiers de log pour la simulation. Tous ces fichiers sont situés dans le dossier `Advanced outputs` du dossier de travail.


* <span style="color:#995E28">osoaa</span> : Nom de fichier log pour la simulation globale. *Valeur par défaut : <span style="color:#009600">"log_osoaa.txt"</span>*.

* <span style="color:#995E28">ang</span> : Nom de fichier log pour le calcul des angles. *Valeur par défaut : <span style="color:#009600">"log_ang.txt"</span>*.

* <span style="color:#995E28">profile</span> : Nom de fichier log pour le calcul des profils atmosphérique et océanique. *Valeur par défaut : <span style="color:#009600">"log_profile.txt"</span>*.

* <span style="color:#995E28">aer</span> : Nom de fichier log pour le calcul des aérosols. *Valeur par défaut : <span style="color:#009600">"log_aer.txt"</span>*.

* <span style="color:#995E28">aermie</span> : Nom de fichier log pour le calcul des aérosols avec la théorie de Mie. *Valeur par défaut : <span style="color:#009600">log_aermie.txt</span>*.

* <span style="color:#995E28">hyd</span> : Nom de fichier log pour le calcul des hydrosols. *Valeur par défaut : <span style="color:#009600">log_hyd.txt</span>*.

* <span style="color:#995E28">hydmie</span> : Nom de fichier log pour le calcul des hydrosols avec la théorie de Mie. *Valeur par défaut : <span style="color:#009600">log_hydmie.txt</span>*.

* <span style="color:#995E28">sea</span> : Nom de fichier log pour le calcul de la surface. *Valeur par défaut : <span style="color:#009600">log_sea.txt</span>*.

* <span style="color:#995E28">sos</span> : Nom de fichier log pour le calcul des SOS. *Valeur par défaut : <span style="color:#009600">log_sos.txt</span>*.

```python
log_files = pyOSOAA.osoaa.LOG()
print("osoaa : ", log_files.osoaa)
print("ang : ", log_files.ang)
print("profile : ", log_files.profile)
print("aer : ", log_files.aer)
print("aermie : ", log_files.aermie)
print("hyd : ", log_files.hyd)
print("hydmie : ", log_files.hydmie)
print("sea : ", log_files.sea)
print("sos : ", log_files.sos)
```

### <span style="color:#D2383F">class RESULTS</span>

Définition des noms de fichiers des résultats de la simulation. Tous ces fichiers sont situés dans le dossier `Advanced outputs` du dossier de travail, sauf les fichiers `vsvza` et `vsz` qui sont situés dans le dossier `Standard_outputs`.


* <span style="color:#995E28">profileatm</span> : Nom de fichier de sortie pour le calcul du profil atmosphérique. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">profilesea</span> : Nom de fichier de sortie pour le calcul du profil océanique. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">aer</span> : Nom de fichier de sortie pour le calcul des aérosols. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">phyto</span> : Nom de fichier de sortie pour le calcul des hydrosols ou particules phytoplancton. Ce calcul inclut la matrice de phase définie par l'utilisateur si elle existe. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">mlp</span> : Nom de fichier de sortie pour le calcul des particules minérales. Les coefficients de la matrice de phase sont nuls si une matrice est définie par l'utilisateur. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">angrad</span> : Nom de fichier de sortie pour la liste des angles utilisés pour les calculs d'intensité et de BRDS/BPDF. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">angmie</span> : Nom de fichier de sortie pour la liste des angles utilisés pour le calcul de la matrice de phase. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">sosbin</span> : Nom de fichier de sortie pour le résultat binaire du calcul des SOS. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">vsvza</span> : Nom de fichier de sortie pour le résultat ascii du calcul des SOS, intensité de sortie en fonction de l'angle zénithal de vue. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">advup</span> : Nom de fichier de sortie pour le résultat ascii du calcul des SOS, intensité de sortie vers le haut en fonction de la profondeur (ou altitude) et en fonction de l'angle zénithal de vue. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">advdown</span> : Nom de fichier de sortie pour le résultat ascii du calcul des SOS, intensité de sortie vers le bas en fonction de la profondeur (ou altitude) et en fonction de l'angle zénithal de vue. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">advphi</span> : Nom de fichier de sortie pour le résultat ascii du calcul des SOS, intensité de sortie vers le haut en fonction de la profondeur (ou altitude), de l'angle zénithal de vue et de l'angle azimutal relatif. Ce fichier n'est disponible qu'en utilisant une version customisée d'OSOAA disponible dans la branche advphi du dépot github. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">vsz</span> : Nom de fichier de sortie pour le résultat ascii du calcul des SOS, intensité de sortie en fonction de la profondeur (ou altitude). *Valeur par défaut : <span style="color:#009600">"resfile_vsz.txt"</span>*.

* <span style="color:#995E28">hydiop</span> : Nom de fichier de sortie pour le calcul des propriétés optiques inhérentes (IOP) du phytoplancton et des sédiments. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

* <span style="color:#995E28">aeriop</span> : Nom de fichier de sortie pour le calcul des propriétés optiques inhérentes (IOP) des aérosols. *Valeur par défaut : <span style="color:#009600">"None"</span>*.

```python
RES = pyOSOAA.osoaa.RESULTS()
print("profileatm : ",RES.profileatm)
print("aer : ",RES.aer)
print("sosbin : ",RES.sosbin)
print("vsz : ",RES.vsz)
```

### <span style="color:#D2383F">class PHYTO</span>

Définition du phytoplancton.



#### <span style="color:#D2383F">class JD</span>

Distribution Junge pour différents modèles.

* <span style="color:#2882E2">mrwa</span> : Partie réelle de l'indice de réfraction des particules de phytoplancton à la longueur d'onde de calcul, mode de particules principal. *Valeur par défaut : <span style="color:#009600">1.05</span>*. *Résolution : 0.001*. *Valeur min : 0*.

* <span style="color:#2882E2">miwa</span> : Partie imaginaire de l'indice de réfraction des particules de phytoplancton à la longueur d'onde de calcul, mode de particules principal. *Valeur par défaut : <span style="color:#009600">0.0</span>*. *Résolution : 0.001*. *Plage de valeurs : [-1.0,0.0]*.

* <span style="color:#2882E2">slope</span> : Pente de la loi de Junge pour les particules de phytoplancton. *Valeur par défaut : <span style="color:#009600">4.0</span>*. *Résolution : 0.001*. *Plage de valeurs : [3.0,5.0], 3 est une valeur singulière*.

* <span style="color:#2882E2">rmin</span> : Rayon minimal de la loi de Junge pour les particules de phytoplancton (µm). *Valeur par défaut : <span style="color:#009600">0.01</span>*. *Résolution : 0.001 µm*. *Valeur min : 0 µm*. *Valeur en mode non expert : <span style="color:#DCA347">0.01</span>*.

* <span style="color:#2882E2">rmax</span> : Rayon maximal de la loi de Junge pour les particules de phytoplancton (µm). *Valeur par défaut : <span style="color:#009600">200</span>*. *Résolution : 0.001 µm*. *Valeur min : 0 µm*. *Valeur en mode non expert : <span style="color:#DCA347">200</span>*.

* <span style="color:#2882E2">rate</span> : Rapport de la loi de Junge dans la distribution globale. *Valeur par défaut : <span style="color:#009600">1.0</span>*. *Résolution : 0.001*. *Plage de valeurs : [0.0,1.0]*.



#### <span style="color:#D2383F">class LND</span>

Distribution log-normale pour différents modèles.

* <span style="color:#2882E2">mrwa</span> : Partie réelle de l'indice de réfraction des particules de phytoplancton à la longueur d'onde de calcul. *Valeur par défaut : <span style="color:#009600">A définir</span>*. *Résolution : 0.001*. *Valeur min : 0*.

* <span style="color:#2882E2">miwa</span> : Partie imaginaire de l'indice de réfraction des particules de phytoplancton à la longueur d'onde de calcul. *Valeur par défaut : <span style="color:#009600">A définir</span>*. *Résolution : 0.001*. *Plage de valeurs : [-1.0,0.0]*.

* <span style="color:#2882E2">sdradius</span> : Rayon modal pour les particules de phytoplancton LND. *Valeur par défaut : <span style="color:#009600">A définir</span>*. *Résolution : 0.001 µm*. *Valeur min : 0 µm*.

* <span style="color:#2882E2">sdvar</span> : Écart-type de la LND des particules de phytoplancton. *Valeur par défaut : <span style="color:#009600">A définir</span>*. *Résolution : 0.001 µm*. *Valeur min : 0 µm*.

* <span style="color:#2882E2">rate</span> : Rapport de la loi LND dans la distribution globale. *Valeur par défaut : <span style="color:#009600">A définir</span>*. *Résolution : 0.001*. *Plage de valeurs : [0.0,1.0]*.



#### <span style="color:#DA70C8">fonction \_\_init\_\_</span>

Définition du profil de phytoplancton.

* <span style="color:#2882E2">chl</span> : Concentration en chlorophylle à la surface de l'océan (mg/m^3). *Valeur par défaut : <span style="color:#009600">0.2</span>*. *Résolution : 0.001 mg/m^3*. *Valeur min : 0 mg/m^3*.

* <span style="color:#2882E2">profiltype</span> : Type de profil de chlorophylle. (1) Profil homogène, (2) Profil gaussien, (3) Profil défini par l'utilisateur. *Valeur par défaut : <span style="color:#009600">1</span>*.

* <span style="color:#2882E2">jd</span> : Distribution en chlorophylle suivant la loi de Junge. *Valeur par défaut : <span style="color:#009600">JD()</span>*.

* <span style="color:#2882E2">sm</span> : Définition du mode secondaire. *Valeur par défaut : <span style="color:#009600">None</span>*.

* <span style="color:#2882E2">tm</span> : Définition du mode tertiaire. *Valeur par défaut : <span style="color:#009600">None</span>*.


#### <span style="color:#DA70C8">fonction SetPrimaryMode</span>

Définition du premier mode de phytoplancton, avec la loi de Junge. Les paramètres sont identiques à ceux de la classe <span style="color:#D2383F">class JD</span>.

* <span style="color:#2882E2">jd</span> : Le premier mode de phytoplancton est calculé grâce à la classe <span style="color:#D2383F">class JD</span>. *Valeur par défaut : <span style="color:#009600">JD(mrwa,miwa,slope,rmin,rmax,rate)</span>*.



#### <span style="color:#DA70C8">fonction SetSecondaryMode</span>

Définition du second mode de phytoplancton, avec une LND. Les paramètres sont identiques à ceux de la classe <span style="color:#D2383F">class LND</span>.

* <span style="color:#2882E2">sm</span> : Le second mode de phytoplancton est calculé grâce à la classe <span style="color:#D2383F">class LND</span>. *Valeur par défaut : <span style="color:#009600">LND(mrwa,miwa,sdradius,sdvar,rate)</span>*.


#### <span style="color:#DA70C8">fonction SetTertiaryMode</span>

Définition du troisième mode de phytoplancton, avec une LND. Les paramètres sont identiques à ceux de la classe <span style="color:#D2383F">class LND</span>.

* <span style="color:#2882E2">tm</span> : Le troisième mode de phytoplancton est calculé grâce à la classe <span style="color:#D2383F">class LND</span>. *Valeur par défaut : <span style="color:#009600">LND(mrwa,miwa,sdradius,sdvar,rate)</span>*.



#### <span style="color:#DA70C8">fonction SetProfilType</span>

Définition du profile de la distribution en phytoplancton, et configuration des paramètres pour chaque type de distribution.

* <span style="color:#2882E2">profiltype</span> : Type de profil de la chlorohpylle. (1) Profil homogène, (2) Profil gaussien, (3) Profil défini par l'utilisateur. *Valeur par défaut : <span style="color:#009600">1</span>*.

* <span style="color:#2882E2">chlbg</span> : Fond de biomasse constant (mg/m^3). *Valeur par défaut : <span style="color:#009600">None</span>*. *Résolution : 0.001 mg/m^3*. *Valeur min : 0 mg/m^3*.

* <span style="color:#2882E2">chlzmax</span> : Concentration maximale en phytoplancton (à la profondeur <span style="color:#2882E2">deep</span>) (mg/m^3). *Valeur par défaut : <span style="color:#009600">A renseigner</span>*. *Résolution : 0.001 mg/m^3*. *Valeur min : 0 mg/m^3*.

* <span style="color:#2882E2">deep</span> : Profondeur maximale du pic gaussien pour le profil de la chlorophylle (m). *Valeur par défaut : <span style="color:#009600">None</span>*. *Résolution : 0.001 m*. *Valeur min : 0 m*.

* <span style="color:#2882E2">width</span> : Largeur du pic gaussien pour le profile de la chlorophylle (m). *Valeur par défaut : <span style="color:#009600">None</span>*. *Résolution : 0.001 m*. *Valeur min : 0 m*.

* <span style="color:#2882E2">userfile</span> : Nom du fichier du profil de phytoplancton défini par l'utilisateur. *Valeur par défaut : <span style="color:#009600">None</span>*.

```python
phyto = pyOSOAA.osoaa.PHYTO()
phyto.profiltype = 1
phyto.jd.rate = 0.8
phyto.SetSecondaryMode(1.05,0,0.58,0.04,0.2)
print("sm.sdradius :", phyto.sm.sdradius)
```
